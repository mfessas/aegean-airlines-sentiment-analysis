Launch the external AirlineRatings scraper and inspect its combined CSV.

Collection requires network access; browser-based scrapers also require Chrome.

## 1. Portable project paths

```python
from pathlib import Path

# Locate the project from the notebook kernel's working directory.
# Start Jupyter inside the project tree; the root name is not significant.
PROJECT_ROOT = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if all((p / name).is_dir() for name in ("Python Code", "Data", "Streamlit"))),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError("Start Jupyter inside the project folder containing Python Code, Data and Streamlit.")

SCRAPING_DIR = PROJECT_ROOT / "Data" / "Scraping"
SOURCE_DIR = SCRAPING_DIR / 'airlineratings_data'
SOURCE_DIR.mkdir(parents=True, exist_ok=True)
```

## Launch scraper

```python
import subprocess
import sys

# The original notebook delegates scraping to a separate Python script.
script = PROJECT_ROOT / "Python Code" / "Scraping" / "airlineratings_multi_airline_paginated_v2.py"
if not script.is_file():
    raise FileNotFoundError(f"Required scraper script not supplied: {script}")

# The legacy script writes airlineratings_data relative to its working directory.
result = subprocess.run(
    [sys.executable, str(script)],
    cwd=SCRAPING_DIR,
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)
print(result.stdout)
print(result.stderr)
result.check_returncode()
```
